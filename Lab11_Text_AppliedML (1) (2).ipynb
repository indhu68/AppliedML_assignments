# Back of Words

We now preprocess the text data with back-of-words processing, which includes tokenizing, filtering of stopwords, and encoding. It is all implemented in CountVectorizer, so let us play with it.



<div id="TODO2"></div>

## Classificaiton

<div id="TODO3"></div>

### TODO3 (5 points)

### TODO4 (5 points)

To make this easy to work with, let us build a pipeline of vectorizer and a classifier.

1. Create a `Pipeline` instance that connects `CountVectorizer` and `MultinomialNB` and store it to `nbclf`.
2. Train `nbclf` with train_data. Use the attributes `data` and `target`.
3. Make a prediciton and store the results in `pred`.
4. Compute the f1 score with `test_data`.

````python
from sklearn.pipeline import Pipeline

# TODO 4.1 - 4.4

nbclf = Pipeline([
    ('vect', CountVectorizer()),
    ('clf', MultinomialNB()),
])
nbclf.fit(train_data.data, train_data.target)
pred = nbclf.predict(test_data.data)
f1Score = f1_score(test_data.target, pred, average='weighted')
f1Score
````

We have very good first results. As the text contains a lot of meaningless words. Let us discard the stop words and see how the performance differs.

## TF-IDF

We can extract a feature representation by using the term frequency and inverse document frequency using the equaion:
$$
 tfidf(w, d) = tf \times \log \Big( \frac{N+1}{N_w + 1} \Big) + 1.
$$
Let us transform the count vector into tf-idf feature representation and see how it affects the performance.

<div id="TODO7"></div>

### TODO7 (5 points)

1. Create `TfidfTransformer` instance and name it `tf_transformer`.
2. `fit_transform` the `X_train_cnt`, store it in `tfidf`, and print the shape.

````python
from sklearn.feature_extraction.text import TfidfTransformer

# TODO 7.1 - 7.2
tf_transformer = TfidfTransformer()
tfidf = tf_transformer.fit_transform(X_train_cnt)


tfidf.shape
````

Well, we are kind of stuck at 0.95 f1 score. ### TODO9 (20 points)

For n-Gram, we first need to decide what n to use. So, we start this practice first searching for the hyperparameters. As we naively selected the alpha for Naive Bayes, we search for it as well using `GridSearchCV`.

The hyperparameter dictionary `parameters` needs to follow the naming convention. The name of an object instance in a pipeline followed by *\"__\"* and the hyperparameter name for each instance. For instance, in the code below, for classifier named 'clf', we want to search for alpha. Thus, the dictionary key is *`clf__alpha`*.

1. Finish the parameters with the list of n-gram ranges and classifier alpha values.
2. Create a pipelined classifier with `CountVectorizer`, `TfidfTransformer`, and `MultinomialNB` and name it `ng_clf`.
2. Create a `GridSearchCV` instance with 5 fold cross validation, name it as `gs_clf`, and train it using train data. Use data and target attributes wherever needed.
3. Print `best_score_` of `gs_clf` object.
4. Print the best parameters in `best_params_`.
5. Print out the `classification_report` with `test_data`.


````python
from sklearn.model_selection import GridSearchCV
# 1)
parameters = {
     'vect__ngram_range': [(1, 1), (1, 2), (2, 2)],
     'clf__alpha':[0.1, 1.0, 3.0, 10.0] ,
}


#2)

ng_clf = Pipeline([
    ('vect', CountVectorizer()),
    ('tfidf', TfidfTransformer()),
    ('clf', MultinomialNB()),
])

````

````python
# 3)
gs_clf = GridSearchCV(ng_clf, parameters, cv=5)
````
